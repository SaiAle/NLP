## **AD)**Create a table displaying relative frequencies with which “modals” (can, could, may, might, will, would, and should) are used in all texts provided in the corpus. 

```python
cfd_1 = nltk.ConditionalFreqDist((genre, words)
for genre in gutenberg.fileids()
for words in gutenberg.words(fileids = genre))
genres = ['austen-emma.txt', 'austen-persuasion.txt', 'austen-sense.txt', 'bible-kjv.txt',
'blake-poems.txt', 'bryant-stories.txt', 'burgess-busterbrown.txt',
'carroll-alice.txt', 'chesterton-ball.txt', 'chesterton-brown.txt',
'chesterton-thursday.txt', 'edgeworth-parents.txt', 'melville-moby_dick.txt',
'milton-paradise.txt', 'shakespeare-caesar.txt', 'shakespeare-hamlet.txt',
'shakespeare-macbeth.txt', 'whitman-leaves.txt']
modals = ['can', 'could', 'may', 'might', 'would', 'will', 'shall']
cfd_1.tabulate(conditions=genres, samples=modals)
```

## **BA)** Choose Kennedy’s speech, using e.g. this code:  

```python
np_words = nltk.corpus.inaugural.words('1961-Kennedy.txt')
```

##**BB)** Identify the 10 most frequently used long words (words longer than 7 characters). 

```python
# Filter's the character
flt_char = ([w.lower() for w in np_words if len(w) > 7])

# Frequency distribution
f_dst = FreqDist(flt_char)
freq = f_dst.most_common(10)

# Creating empty arrays
e_1 = []
e_2 = []

# To Seperate the words
for i in range(10):
    e_1.append(freq[i][0])

for i in range(10):
    e_2.append(freq[i][1])

# To Create dataframe 
df_freq = pd.DataFrame({'Most Frequently Used Top_10':e_1, 'Repeated No.Of Times':e_2})
```

```python
df_freq
```

Here are the top 10 words used most frequently which has each word more than seven character's.


##**BC)** Which one of those 10 words has the largest number of synonyms? Use WordNet as a helper:

```python
# Creating an array
np_syn = []

# Loop trough the distribution
for i in f_dst.most_common(10):
    
    print ('Word: ', i[0])
    count=0
    
    # Print synonym
    print ('Synonymns: ')
    for j in wn.synsets(i[0]):
        print ('  ',j.lemma_names())
        count+=len(j.lemma_names())
    np_syn.append([i[0], count])  
```

Using wordnet as a helper to print the synonyms for the top 10 words which have more than seven characters. Started with creating an empty array and used for loop to find the most common 10 words and created another for loop inside the for loop to print the synonym for each top word.

```python
# Creaty empty arrays
e_1 = []
e_2 = []

# Get synonyms
for i in range(10):
    e_1.append(np_syn[i][0])

for j in range(10):
    e_2.append(np_syn[j][1])

# Create dataframe
df_sn = pd.DataFrame({'Most Frequently Used Top_10':e_1, 'No.Of synonyms':e_2})
```

```python
df_sn.sort_values(by="No.Of synonyms", ascending=False)
```

Listing all the most used words with the total number of synonyms and sorting in asceding order.

```python
# Create array
a_hyp = []

# Loop trough distribution
for i in f_dst.most_common(10):
    # Print word
    print ('Word: ', i[0])
    count=0
    
    # Print Hyponyms
    print ('Hyponyms: ')
    for j in wn.synsets(i[0]):
        print (j)
        print (j.hyponyms())
        count+= len(j.hyponyms())
    a_hyp.append([i[0], count])
```
